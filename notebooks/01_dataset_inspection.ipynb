```python
from collections import Counter

all_models = Counter()

for row in dataset:
    all_models[row["model_a"]] += 1
    all_models[row["model_b"]] += 1

print("Number of unique models:", len(all_models))

for model, count in all_models.most_common():
    print(f"{model:45s} {count}")
```

```python
from collections import Counter

single_turn_models = Counter()

for row in dataset:
    if len(row["conversation_a"]) == 2:
        single_turn_models[row["model_a"]] += 1
        single_turn_models[row["model_b"]] += 1

for model, count in single_turn_models.most_common():
    print(f"{model:45s} {count}")
```

```python
def get_family(model_name):
    name = model_name.lower()

    if "gpt" in name or "chatgpt" in name:
        return "GPT"

    if "claude" in name:
        return "Claude"

    if "gemini" in name:
        return "Gemini"

    return None


family_counts = Counter()

for model, count in single_turn_models.items():
    family = get_family(model)

    if family is not None:
        family_counts[family] += count

print(family_counts)
```

```python
from collections import Counter

family_language_counts = Counter()

for row in dataset:
    if len(row["conversation_a"]) != 2:
        continue

    for model_key in ["model_a", "model_b"]:
        model = row[model_key]
        family = get_family(model)

        if family is not None:
            language = row["language"]
            family_language_counts[(family, language)] += 1

for key, count in family_language_counts.most_common():
    print(key, count)
```

```python
english_counts = {
    family: family_language_counts[(family, "en")]
    for family in ["GPT", "Claude", "Gemini"]
}

print(english_counts)
```

```python
from collections import Counter

domain_counts = Counter()

for row in dataset:

    # Keep only single-turn conversations
    if len(row["conversation_a"]) != 2:
        continue

    # Keep only English
    if row["language"] != "en":
        continue

    tag = row["category_tag"] or {}

    is_math = tag.get("math_v0.1", {}).get("math", False)

    is_creative = tag.get(
        "creative_writing_v0.1", {}
    ).get("creative_writing", False)

    is_code = bool(row["is_code"])

    # Check both model A and model B
    for model_key in ["model_a", "model_b"]:

        family = get_family(row[model_key])

        if family not in ["GPT", "Claude", "Gemini"]:
            continue

        if is_code:
            domain_counts[(family, "code")] += 1

        if is_math:
            domain_counts[(family, "math")] += 1

        if is_creative:
            domain_counts[(family, "creative_writing")] += 1


for family in ["GPT", "Claude", "Gemini"]:

    print(f"\n{family}")

    for domain in ["code", "math", "creative_writing"]:
        print(domain, domain_counts[(family, domain)])
```

```python
def extract_text(message):
    text_parts = []

    for item in message["content"]:
        if item.get("type") == "text" and item.get("text") is not None:
            text_parts.append(item["text"])

    return "\n".join(text_parts)
```

```python
print("Prompt:")
print(extract_text(dataset[0]["conversation_a"][0]))

print("\nResponse:")
print(extract_text(dataset[0]["conversation_a"][1])[:1000])
```

```python
import pandas as pd

rows = []

for row in dataset:

    # 1. Single-turn only
    if len(row["conversation_a"]) != 2:
        continue

    # 2. English only
    if row["language"] != "en":
        continue

    # Get useful category information for RQ3
    tag = row["category_tag"] or {}

    is_math = tag.get(
        "math_v0.1", {}
    ).get("math", False)

    is_creative = tag.get(
        "creative_writing_v0.1", {}
    ).get("creative_writing", False)

    is_code = bool(row["is_code"])

    # Process model A and model B separately
    for side in ["a", "b"]:

        model = row[f"model_{side}"]
        family = get_family(model)

        # Keep only our 3 target families
        if family not in ["GPT", "Claude", "Gemini"]:
            continue

        conversation = row[f"conversation_{side}"]

        prompt = extract_text(conversation[0])
        response = extract_text(conversation[1])

        rows.append({
            "group_id": row["id"],
            "family": family,
            "model": model,
            "prompt": prompt,
            "response": response,
            "language": row["language"],
            "is_code": is_code,
            "is_math": is_math,
            "is_creative_writing": is_creative
        })


candidate_df = pd.DataFrame(rows)
```

```python
import hashlib

def make_prompt_group_id(prompt):
    return hashlib.sha1(
        prompt.encode("utf-8")
    ).hexdigest()

candidate_df["prompt_group_id"] = (
    candidate_df["prompt"].apply(make_prompt_group_id)
)
```

```python
import re

def has_direct_identity_leak(row):
    response = row["response"][:500].lower()
    family = row["family"]

    if family == "GPT":
        patterns = [
            r"\bi am chatgpt\b",
            r"\bi'm chatgpt\b",
            r"\bas chatgpt\b",
            r"\bi am an openai",
            r"\bi'm an openai"
        ]

    elif family == "Claude":
        patterns = [
            r"\bi am claude\b",
            r"\bi'm claude\b",
            r"\bas claude\b",
            r"\bwith claude, an ai assistant\b",
            r"\bi \(claude\) work\b"
        ]

    elif family == "Gemini":
        patterns = [
            r"\bi am gemini\b",
            r"\bi'm gemini\b",
            r"\bas gemini\b"
        ]

    else:
        return False

    return any(re.search(pattern, response) for pattern in patterns)


candidate_df["direct_identity_leak"] = candidate_df.apply(
    has_direct_identity_leak,
    axis=1
)

print("Likely direct identity leakage:")
print(candidate_df["direct_identity_leak"].value_counts())

print("\nBy family:")
print(
    candidate_df[
        candidate_df["direct_identity_leak"]
    ]["family"].value_counts()
)
```

```python
def assign_domain(row):
    code = bool(row["is_code"])
    math = bool(row["is_math"])
    creative = bool(row["is_creative_writing"])

    total = code + math + creative

    # More than one category
    if total > 1:
        return "mixed"

    if code:
        return "code"

    if math:
        return "math"

    if creative:
        return "creative_writing"

    return "general"


selected_df["domain"] = selected_df.apply(
    assign_domain,
    axis=1
)
```

```python
import pandas as pd

SEED = 42

domain_targets = {
    "general": 800,
    "code": 500,
    "math": 350,
    "creative_writing": 350
}

families = ["GPT", "Claude", "Gemini"]

sampled_parts = []

for family in families:
    for domain, n in domain_targets.items():

        subset = selected_df[
            (selected_df["family"] == family) &
            (selected_df["domain"] == domain)
        ]

        sampled = subset.sample(
            n=n,
            random_state=SEED
        )

        sampled_parts.append(sampled)

final_df = pd.concat(
    sampled_parts,
    ignore_index=True
)

# Shuffle the finished dataset
final_df = final_df.sample(
    frac=1,
    random_state=SEED
).reset_index(drop=True)

# Give every response its own unique ID
final_df.insert(
    0,
    "sample_id",
    range(1, len(final_df) + 1)
)

print("Final dataset size:", len(final_df))
```

```python
from sklearn.model_selection import GroupShuffleSplit

SEED = 42

# First split: 70% train, 30% temporary
split1 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.30,
    random_state=SEED
)

train_idx, temp_idx = next(
    split1.split(
        final_df,
        groups=final_df["prompt_group_id"]
    )
)

train_df = final_df.iloc[train_idx].copy()
temp_df = final_df.iloc[temp_idx].copy()


# Second split: divide temporary set equally
# 15% validation + 15% test
split2 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.50,
    random_state=SEED
)

val_idx, test_idx = next(
    split2.split(
        temp_df,
        groups=temp_df["prompt_group_id"]
    )
)

val_df = temp_df.iloc[val_idx].copy()
test_df = temp_df.iloc[test_idx].copy()
```

```python
train_df["split"] = "train"
val_df["split"] = "validation"
test_df["split"] = "test"

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
```
